```python
from mytorch.train import *
from mytorch.HT import *
from mytorch.TT import *
```

```python
losses_tt, accs_tt = train_tt()
```

```python
torch.manual_seed(0)
losses_ht, accs_ht = train_ht()
```

```python
def plot_ht_tt(losses_ht, accs_ht, losses_tt, accs_tt):
    epochs = range(1, len(losses_ht) + 1)

    fig, axes = plt.subplots(1, 2, figsize=(10, 4))

    axes[0].plot(epochs, losses_ht, label="HT", linestyle="-", color="black")
    axes[0].plot(epochs, losses_tt, label="TT", linestyle=":", color="black")
    axes[0].set_title("Training Loss")
    axes[0].set_xlabel("Epoch", fontsize=15)
    axes[0].set_ylabel("Loss", fontsize=15)
    axes[0].legend(fontsize=15)
  

    axes[1].plot(epochs, accs_ht, label="HT", linestyle="-", color="black")
    axes[1].plot(epochs, accs_tt, label="TT", linestyle=":", color="black")
    axes[1].set_title("Training Accuracy")
    axes[1].set_xlabel("Epoch", fontsize=15)
    axes[1].set_ylabel("Accuracy (%)", fontsize=15)
    axes[1].legend(fontsize=15)


    plt.tight_layout()
    plt.show()

plot_ht_tt(losses_ht, accs_ht, losses_tt, accs_tt)
```
